Corporate insolvency is an important indicator of financial distress in the business sector. Changes in insolvency activity can provide useful signals for credit risk monitoring, particularly for financial institutions, trade credit providers, and businesses exposed to corporate customers.

However, a high number of insolvency appointments does not necessarily indicate worsening industry conditions. Appointment activity may reflect long-term growth, short-term fluctuations, changes in industry concentration, or exceptional events involving multiple companies.

This project uses publicly available corporate insolvency data from the Australian Securities and Investments Commission (ASIC), covering 58,115 appointment records across 64 reporting months from July 2021 to October 2026.

**Primary Objective:** Develop an evidence-based industry distress monitoring framework that combines historical and recent evidence to support credit risk prioritisation without treating insolvency counts as company-level default probabilities.

## 1. Business Understanding

### 1.1 Business Problem

Financial institutions, trade credit providers, and businesses with corporate customers need to monitor financial distress across industries to manage potential credit exposure.

However, relying solely on insolvency counts can lead to misleading conclusions:

- **Industry size differences:** Industries with more companies may naturally record more insolvency cases.
- **Short-term volatility:** A temporary increase in insolvency activity may not represent sustained deterioration.
- **Event concentration:** A large number of insolvency appointments may occur within a concentrated group of companies or events.
- **Historical context:** Current insolvency levels may appear concerning without comparison to previous periods.

These limitations make it difficult for credit risk teams to distinguish persistent industry distress from temporary or concentrated changes in insolvency activity.

### 3.1 Monthly Insolvency Trend & Moving Average

This analysis examines monthly ASIC Series 1 appointment counts from July 2021 to October 2026.

Monthly counts are displayed alongside a trailing three-month moving average (MA3) to distinguish short-term fluctuations from broader movements in insolvency activity.

**Methodology**

The MA3 is calculated as the arithmetic mean of the current month and the two preceding months.

\[
MA3_t = \frac{X_t + X_{t\text{-}1} + X_{t\text{-}2}}{3}
\]

Where:
- \(X_t\) represents the appointment count in month \(t\).
- \(MA3_t\) represents the three-month moving average.

The first two months do not have sufficient observations to calculate MA3.

**Analytical Purpose:** Establish a national trend baseline before examining year-over-year changes and industry-specific distress signals.

**Interpretation Note:** MA3 smooths monthly fluctuations but does not remove seasonality or explain the causes of observed changes.

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Consistent portfolio chart colours
NAVY = "#243F5B"
SLATE = "#71869B"
ORANGE = "#E58A35"
GRID = "#E8EDF2"

# Prepare monthly trend data
trend_df = monthly_df.sort_values("reporting_month").copy()

# Calculate the 3-month moving average
trend_df["ma3"] = (
    trend_df["insolvency_count"]
    .rolling(window=3, min_periods=3)
    .mean()
)

# Create chart
fig, ax = plt.subplots(figsize=(12, 5.5))

fig.patch.set_facecolor("white")
ax.set_facecolor("white")

# Monthly insolvency appointments
ax.plot(
    trend_df["reporting_month"],
    trend_df["insolvency_count"],
    color=SLATE,
    linewidth=1.5,
    alpha=0.85,
    label="Monthly appointments"
)

# 3-month moving average
ax.plot(
    trend_df["reporting_month"],
    trend_df["ma3"],
    color=ORANGE,
    linewidth=2.8,
    label="3-month moving average"
)

# Chart title
ax.set_title(
    "Australian Corporate Insolvency | Monthly Trend & MA3",
    fontsize=13,
    fontweight="bold",
    color=NAVY,
    pad=18
)

# Axis labels
ax.set_xlabel(
    "Reporting Month",
    fontsize=9,
    color=NAVY,
    labelpad=10
)

ax.set_ylabel(
    "Number of Series 1 Appointments",
    fontsize=9,
    color=NAVY,
    labelpad=10
)

# Format reporting years
ax.xaxis.set_major_locator(mdates.YearLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))

# Set the horizontal axis to the observed reporting period
ax.set_xlim(
    trend_df["reporting_month"].min(),
    trend_df["reporting_month"].max() + pd.Timedelta(days=15)
)

# Light horizontal gridlines
ax.grid(
    axis="y",
    color=GRID,
    linewidth=0.8
)

ax.set_axisbelow(True)

# Remove chart borders
for spine in ax.spines.values():
    spine.set_visible(False)

# Format tick labels
ax.tick_params(
    axis="both",
    colors=NAVY,
    labelsize=9,
    length=0,
    pad=8
)

# Legend
ax.legend(
    loc="upper left",
    frameon=False,
    fontsize=9,
    labelcolor=NAVY
)

plt.tight_layout()
plt.show()
```

**Key Findings - Monthly Trend & Moving Average**

1. **Sustained upward trend:** Australian corporate insolvency activity increased substantially between 2021 and 2026. The three-month moving average confirms that the increase extended beyond isolated monthly spikes.

2. **Significant monthly volatility:** Monthly appointment counts fluctuate considerably, particularly during 2024-2026. The moving average helps distinguish these short-term movements from the broader trend.

3. **Elevated recent activity:** Insolvency counts remained high toward October 2026 relative to earlier years, although the moving average eased after a recent peak.

**Business Implication:** The moving average provides a clearer view of the underlying direction of appointment activity and helps distinguish sustained changes from short-term volatility.

### 3.2 Year-over-Year Growth

Year-over-year (YoY) growth measures the percentage change in corporate insolvency counts compared with the same month in the previous year.

Unlike month-over-month comparisons, YoY analysis compares equivalent calendar months and therefore reduces distortions associated with recurring seasonal patterns.

**Methodology:**

\[
YoY_t = \left(\frac{X_t}{X_{t\text{-}12}} \text{-} 1\right) \times 100
\]

Where:
- \(X_t\) is the insolvency count in the current month.
- \(X_{t\text{-}12}\) is the insolvency count in the same month of the previous year.

The first 12 months do not have a prior-year comparison within the dataset and are therefore excluded from YoY calculations.

```python
# Create a working copy of the monthly trend dataset
yoy_df = monthly_df.copy()

# Get the insolvency count from the same month last year
yoy_df["previous_year_count"] = (
    yoy_df["insolvency_count"].shift(12)
)

# Calculate year-over-year percentage growth
yoy_df["yoy_growth_pct"] = (
    (
        yoy_df["insolvency_count"]
        / yoy_df["previous_year_count"]
        - 1
    ) * 100
).round(2)

# Display the latest 12 months
display(yoy_df.tail(12))
```

**Key Findings - Year-over-Year Growth**

1. **Mixed growth signals in 2026:** Six of the first ten months of 2026 recorded lower insolvency counts than the corresponding months in 2025, indicating that the elevated level of insolvency activity was not accompanied by consistent year-over-year growth.

2. **Exceptional August increase:** August 2026 recorded 1,808 Series 1 cases, representing a 65.87% increase from August 2025. This warrants further investigation into industry-level contributions.

3. **October decline:** October 2026 recorded 1,090 cases, down 26.35% from October 2025. This decline should not be interpreted as evidence of a sustained improvement without examining a longer period.

**Business Implication:** Monthly year-over-year comparisons help identify exceptional changes that warrant investigation, particularly the August 2026 increase.

### 3.3 Year-to-Date Insolvency Comparison

Monthly year-over-year growth can be influenced by unusually high or low observations in individual months.

To evaluate the broader direction of insolvency activity, a year-to-date (YTD) comparison is performed using the same January–October reporting window for 2025 and 2026.

**Methodology:**

\[
YTD\ Growth = \left(\frac{YTD_{2026}}{YTD_{2025}} \text{-} 1\right) \times 100
\]

Comparing equivalent ten-month periods avoids distortions caused by comparing a partial year with a complete year.

The analysis evaluates whether overall insolvency activity has increased or decreased despite substantial monthly volatility.

```python
# Select January to October for 2025 and 2026
ytd_df = monthly_df[
    (monthly_df["reporting_month"].dt.year.isin([2025, 2026])) &
    (monthly_df["reporting_month"].dt.month <= 10)
].copy()

# Calculate total insolvency counts for each year
ytd_summary = (
    ytd_df
    .groupby(ytd_df["reporting_month"].dt.year)["insolvency_count"]
    .sum()
)

# Calculate year-to-date growth
ytd_growth = (
    (ytd_summary.loc[2026] / ytd_summary.loc[2025]) - 1
) * 100

print("January–October 2025:", ytd_summary.loc[2025])
print("January–October 2026:", ytd_summary.loc[2026])
print("YTD Growth (%):", round(ytd_growth, 2))
```

**Key Findings - Year-to-Date Comparison**

- **Slight decline in 2026:** Australia recorded 12,073 Series 1 cases during January-October 2026, compared with 12,272 during the equivalent period in 2025, representing a decrease of 199 cases (-1.62%).

- **Monthly spikes do not define the overall trend:** Despite a 65.87% year-over-year increase in August 2026, cumulative insolvency activity was marginally lower than in the corresponding 2025 period.

- **Elevated activity without continued aggregate growth:** Insolvency counts remained high relative to earlier years, but the year-to-date comparison does not indicate further nationwide growth in 2026.

**Business Implication:** The January-October comparison provides a consistent basis for evaluating the latest national position without allowing a single exceptional month to dominate the overall assessment.

### 3.4 Seasonality Analysis

This analysis examines whether corporate insolvency appointments display recurring patterns across calendar months.

To reduce the influence of long-term growth in overall appointment volumes, monthly counts are expressed as a percentage of their respective financial-year totals.

Only complete financial years (FY2021-22 through FY2025-26) are included, ensuring that all twelve calendar months are represented in each year.

The analysis is descriptive and does not assume that observed monthly differences are caused by seasonality.

```python
# Select the five completed financial years
seasonality_df = clean_df[
    clean_df["financial_year_start"].between(2021, 2025)
].copy()

# Count appointments by financial year and calendar month
seasonality_monthly = (
    seasonality_df
    .groupby(["financial_year_start", "calendar_month"])
    .size()
    .reset_index(name="insolvency_count")
)

# Calculate total appointments for each financial year
seasonality_monthly["fy_total"] = (
    seasonality_monthly
    .groupby("financial_year_start")["insolvency_count"]
    .transform("sum")
)

# Calculate each month's share of its financial-year total
seasonality_monthly["monthly_share_pct"] = (
    seasonality_monthly["insolvency_count"]
    / seasonality_monthly["fy_total"] * 100
)

# Calculate average monthly share across completed financial years
seasonality_summary = (
    seasonality_monthly
    .groupby("calendar_month")["monthly_share_pct"]
    .agg(["mean", "min", "max"])
    .round(2)
    .reset_index()
)

display(seasonality_summary)
```

**Key Findings - Seasonality Analysis**

1. **Consistently low January activity:** January accounts for an average of 4.86% of annual insolvency appointments across the five completed financial years, substantially below the equal-month benchmark of 8.33%. Its relatively narrow historical range suggests a recurring monthly pattern.

2. **Higher activity during March-June:** March (9.85%), May (10.06%), and June (9.76%) record above-average shares of annual appointments, suggesting that insolvency activity is not evenly distributed throughout the year.

3. **August is not historically a high-activity month:** August represents an average of 7.97% of annual appointments across the completed financial years. The exceptional increase in August 2026 should therefore be investigated rather than assumed to reflect normal seasonal behaviour.

**Business Implication:** Monthly comparisons should account for recurring calendar patterns. Year-over-year comparisons using equivalent months provide a more appropriate basis for identifying unusual changes than comparisons between adjacent months.

**Limitation:** These findings are descriptive and based on five completed financial years. They do not establish the underlying causes of the observed monthly patterns.

High appointment concentration does not independently indicate a high insolvency rate because industry company populations differ.

### 4.2 Multi-Year Industry Trends

This analysis examines how corporate insolvency appointment counts have evolved across Australian industries over multiple years.

To ensure comparability, January-October observations are selected for each calendar year from 2022 to 2026. The equivalent period in 2021 is excluded because the dataset begins in July 2021.

The analysis aims to:

- Identify industries experiencing sustained multi-year increases in appointment activity.
- Distinguish gradual increases from temporary fluctuations and recent reversals.
- Establish historical context for the subsequent 2025-2026 year-over-year comparison.

This multi-year perspective helps determine whether recent industry movements represent a continuation of historical trends or a change in direction.

```python
# Select comparable January–October periods from 2022 to 2026
multi_year_df = clean_df[
    (clean_df["calendar_year"].between(2022, 2026)) &
    (clean_df["calendar_month"] <= 10)
].copy()

# Count appointments by industry and calendar year
multi_year_counts = (
    multi_year_df
    .groupby(["Industry type (division)", "calendar_year"])
    .size()
    .unstack(fill_value=0)
)

# Ensure all industries and comparison years are represented
multi_year_counts = multi_year_counts.reindex(
    index=industry_summary["Industry type (division)"],
    columns=[2022, 2023, 2024, 2025, 2026],
    fill_value=0
)

# Validate the aggregated appointment counts
assert multi_year_counts.to_numpy().sum() == len(multi_year_df)

# Sort industries by their latest appointment counts
multi_year_counts = multi_year_counts.sort_values(
    by=2026,
    ascending=False
)

# Display the multi-year industry comparison
display(multi_year_counts.reset_index())
```

```python
# Create a working dataset for multi-year industry indicators
multi_year_metrics = multi_year_counts.copy()

# Calculate absolute change between 2022 and 2026
multi_year_metrics["long_term_change"] = (
    multi_year_counts[2026] - multi_year_counts[2022]
)

# Calculate percentage change between 2022 and 2026
multi_year_metrics["long_term_growth_pct"] = (
    multi_year_metrics["long_term_change"]
    / multi_year_counts[2022].replace(0, float("nan"))
    * 100
)

# Calculate annual changes between consecutive years
annual_changes = multi_year_counts.diff(axis=1)

# Count the number of annual increases
multi_year_metrics["positive_growth_years"] = (
    annual_changes[[2023, 2024, 2025, 2026]] > 0
).sum(axis=1)

# Calculate the percentage of annual periods with growth
multi_year_metrics["annual_growth_consistency_pct"] = (
    multi_year_metrics["positive_growth_years"] / 4 * 100
)

# Calculate the latest year-over-year growth
multi_year_metrics["latest_yoy_growth_pct"] = (
    (multi_year_counts[2026] - multi_year_counts[2025])
    / multi_year_counts[2025].replace(0, float("nan"))
    * 100
)

# Exclude Unknown from industry-level comparison
multi_year_metrics = multi_year_metrics.drop(
    index="Unknown",
    errors="ignore"
)

# Sort industries by long-term absolute change
multi_year_metrics = multi_year_metrics.sort_values(
    by="long_term_change",
    ascending=False
)

# Display the multi-year industry indicators
display(multi_year_metrics.round(2).reset_index())
```

```python
# Calculate national appointment counts for each year
# Include Unknown to preserve official national totals
national_multi_year_counts = multi_year_counts.sum(axis=0)

# Calculate national long-term growth between 2022 and 2026
national_long_term_growth_pct = (
    (national_multi_year_counts[2026]
     - national_multi_year_counts[2022])
    / national_multi_year_counts[2022]
    * 100
)

# Calculate industry growth relative to the national benchmark
multi_year_metrics["growth_vs_national_pp"] = (
    multi_year_metrics["long_term_growth_pct"]
    - national_long_term_growth_pct
)

# Display national appointment counts
display(
    national_multi_year_counts
    .rename("national_appointments")
    .reset_index()
)

# Display national long-term growth
print(
    f"National long-term growth: "
    f"{national_long_term_growth_pct:.2f}%"
)

# Display industry growth relative to the national benchmark
display(
    multi_year_metrics[
        [
            "long_term_change",
            "long_term_growth_pct",
            "growth_vs_national_pp",
            "positive_growth_years",
            "latest_yoy_growth_pct"
        ]
    ]
    .sort_values("growth_vs_national_pp", ascending=False)
    .round(2)
    .reset_index()
)
```

Each rolling observation represents the total number of appointments recorded during the preceding 12 months, including the reporting month.

The analysis aims to:

```python
# Exclude Unknown from industry-level trend analysis
rolling_industry_counts = rolling_12m_counts.drop(
    columns="Unknown"
)

# Create a working dataset for rolling trend indicators
rolling_trend_metrics = pd.DataFrame(
    index=rolling_industry_counts.columns
)

# Get the first and latest rolling 12-month counts
rolling_trend_metrics["initial_12m_count"] = (
    rolling_industry_counts.iloc[0]
)

rolling_trend_metrics["latest_12m_count"] = (
    rolling_industry_counts.iloc[-1]
)

# Identify the highest rolling count in the observation period
rolling_trend_metrics["peak_12m_count"] = (
    rolling_industry_counts.max()
)

# Identify the most recent month reaching the observed peak
rolling_trend_metrics["peak_month"] = (
    rolling_industry_counts.iloc[::-1]
    .idxmax()
    .dt.strftime("%b %Y")
)

# Calculate the percentage change from the observed peak
rolling_trend_metrics["change_from_peak_pct"] = (
    (
        rolling_trend_metrics["latest_12m_count"]
        - rolling_trend_metrics["peak_12m_count"]
    )
    / rolling_trend_metrics["peak_12m_count"]
    .replace(0, float("nan"))
    * 100
)

# Calculate the change in rolling totals over the latest six months
rolling_trend_metrics["rolling_6m_change_pct"] = (
    (
        rolling_industry_counts.iloc[-1]
        - rolling_industry_counts.iloc[-7]
    )
    / rolling_industry_counts.iloc[-7]
    .replace(0, float("nan"))
    * 100
)

# Calculate the change in rolling totals over the latest 12 months
rolling_trend_metrics["rolling_12m_change_pct"] = (
    (
        rolling_industry_counts.iloc[-1]
        - rolling_industry_counts.iloc[-13]
    )
    / rolling_industry_counts.iloc[-13]
    .replace(0, float("nan"))
    * 100
)

# Validate industry coverage
assert len(rolling_trend_metrics) == 19

# Validate changes relative to observed peaks
assert (
    rolling_trend_metrics["change_from_peak_pct"] <= 0
).all()

# Sort industries by their latest rolling appointment counts
rolling_trend_metrics = rolling_trend_metrics.sort_values(
    "latest_12m_count",
    ascending=False
)

# Display rolling trend indicators
display(rolling_trend_metrics.round(2).reset_index())
```

### 4.4 Historical Industry Share Evolution

This analysis examines how the distribution of corporate insolvency appointments across Australian industries has changed over time.

Industry shares are calculated using rolling 12-month appointment counts relative to national rolling appointment totals, covering 53 complete rolling periods from June 2022 to October 2026.

### 4.6 Historical Deviation Analysis

This analysis evaluates whether industry-level insolvency appointment counts in January-October 2026 are elevated relative to historical observations.

To ensure comparability, the analysis uses January-October observations for each year from 2022 to 2026.

**Methodology:**

- **Historical Baseline:** Average January-October appointment count across 2022-2025.
- **Current Activity:** January-October 2026 appointment count.
- **Historical Deviation (%):** Percentage difference between current activity and the historical baseline.

\[
Historical\ Deviation = \left(\frac{Count_{2026}}{Historical\ Average} \text{-} 1\right) \times 100
\]

A positive deviation indicates that current appointment counts exceed the historical average.

**Interpretation:** Historical deviation measures how elevated current activity is relative to earlier years. It does not establish statistical abnormality, causation, or company-level insolvency risk.

Because national insolvency activity increased over the historical period, positive deviations may partly reflect broader national trends rather than industry-specific deterioration.

```python
# Select comparable January–October periods, 2022–2026
historical_df = clean_df[
    (clean_df["calendar_year"].between(2022, 2026)) &
    (clean_df["calendar_month"] <= 10)
].copy()

# Count appointments by industry and calendar year
historical_counts = (
    historical_df
    .groupby(["Industry type (division)", "calendar_year"])
    .size()
    .unstack(fill_value=0)
)

# Ensure all comparison years are represented
historical_counts = historical_counts.reindex(
    columns=[2022, 2023, 2024, 2025, 2026],
    fill_value=0
)

# Calculate the historical baseline
historical_counts["historical_average"] = (
    historical_counts[[2022, 2023, 2024, 2025]]
    .mean(axis=1)
)

# Calculate the deviation from historical average
historical_counts["historical_deviation_pct"] = (
    (
        historical_counts[2026]
        / historical_counts["historical_average"].replace(0, float("nan"))
        - 1
    ) * 100
).round(2)

# Calculate absolute deviation
historical_counts["absolute_deviation"] = (
    historical_counts[2026]
    - historical_counts["historical_average"]
).round(2)

# Sort by absolute deviation
historical_counts = historical_counts.sort_values(
    "absolute_deviation",
    ascending=False
)

# Display the results
display(historical_counts.reset_index())
```

**Key Findings - Historical Deviation**

1. **Construction recorded the largest absolute deviation:** January-October 2026 appointments exceeded the 2022-2025 historical average by 1,021.75 cases (+43.33%).

2. **Health Care and Mining showed substantial relative deviations:** Health Care and Social Assistance (+82.19%) and Mining (+64.30%) recorded elevated counts relative to their historical averages. However, these deviations partly reflect increases over the historical period.

3. **Historical elevation and recent growth are different signals:** Accommodation and Food Services remained 15.15% above its historical average despite declining 17.16% compared with January-October 2025.

**Business Implication:** Historical deviation provides context for current appointment volumes but should not be interpreted independently as evidence of accelerating financial distress.

### 4.7 Recent Industry Share Shift

This analysis examines changes in each industry's contribution to national Series 1 insolvency appointments between January-October 2025 and January-October 2026.

Industry shares are calculated using the total national appointment count in each comparable period.

**Methodology:**

\[
Industry\ Share = \frac{Industry\ Appointments}{National\ Appointments}\times100
\]

\[
Share\ Shift = Share_{2026}\ \text{-}\ Share_{2025}
\]

Share shifts are expressed in percentage points (pp), not percentage growth.

A positive share shift indicates that an industry accounts for a larger proportion of national appointments than in the previous year.

**Interpretation:** An increasing share may indicate a growing concentration of national insolvency activity in a particular industry. It does not measure the probability of insolvency among companies operating in that industry.

```python
# Create a working copy of the industry YoY results
share_shift_df = industry_yearly.copy()

# Calculate national totals for each comparison year
national_2025 = share_shift_df["count_2025"].sum()
national_2026 = share_shift_df["count_2026"].sum()

# Calculate industry shares
share_shift_df["share_2025_pct"] = (
    share_shift_df["count_2025"] / national_2025 * 100
)

share_shift_df["share_2026_pct"] = (
    share_shift_df["count_2026"] / national_2026 * 100
)

# Calculate the change in percentage points
share_shift_df["share_shift_pp"] = (
    share_shift_df["share_2026_pct"]
    - share_shift_df["share_2025_pct"]
).round(2)

# Round shares for presentation
share_shift_df["share_2025_pct"] = (
    share_shift_df["share_2025_pct"].round(2)
)

share_shift_df["share_2026_pct"] = (
    share_shift_df["share_2026_pct"].round(2)
)

# Rank industries by share increase
share_shift_df = share_shift_df.sort_values(
    "share_shift_pp",
    ascending=False
)

display(
    share_shift_df[
        [
            "count_2025",
            "count_2026",
            "share_2025_pct",
            "share_2026_pct",
            "share_shift_pp"
        ]
    ].reset_index()
)
```

1. **Construction recorded the largest share increase:** Its contribution to national Series 1 appointments rose from 24.40% in January-October 2025 to 28.00% in January-October 2026, an increase of 3.60 percentage points.

2. **Smaller share gains in Mining and Health Care:** Mining (+0.30 pp) and Health Care and Social Assistance (+0.29 pp) also accounted for larger proportions of national appointments.

3. **Accommodation and Food Services recorded the largest share decline:** Its share decreased from 15.91% to 13.39% (-2.51 pp), consistent with its year-over-year decline in appointment counts.

**Business Implication:** Construction exhibits the clearest increase in the concentration of national insolvency appointments. This strengthens the case for prioritising the sector for further monitoring, while other industries require additional evidence of sustained deterioration.

### 4.8 Persistence of Industry-Level Increases

This analysis evaluates whether industry-level increases in insolvency appointments were sustained across multiple months or concentrated in isolated periods.

For each industry, monthly Series 1 appointment counts in January-October 2026 are compared with the corresponding months in 2025.

**Persistence Indicator:**

\[
Persistence = \frac{\text{Months with positive YoY count changes}}{10}\times100
\]

A month is classified as an increase only when the 2026 count exceeds the count for the same month in 2025.

A higher persistence percentage indicates more frequent year-over-year increases, but does not necessarily imply larger increases or greater company-level insolvency risk.

**Limitation:** The indicator measures the frequency of positive monthly changes, not their magnitude or statistical significance.

```python
# Select January-October observations for 2025 and 2026
persistence_df = clean_df[
    (clean_df["calendar_year"].isin([2025, 2026])) &
    (clean_df["calendar_month"] <= 10)
].copy()

# Count appointments by industry, month and year
monthly_industry_counts = (
    persistence_df
    .groupby([
        "Industry type (division)",
        "calendar_month",
        "calendar_year"
    ])
    .size()
    .unstack(fill_value=0)
)

# Ensure both comparison years exist
monthly_industry_counts = monthly_industry_counts.reindex(
    columns=[2025, 2026],
    fill_value=0
)

# Create all industry-month combinations
industries = clean_df["Industry type (division)"].unique()
months = range(1, 11)

complete_index = pd.MultiIndex.from_product(
    [industries, months],
    names=["Industry type (division)", "calendar_month"]
)

# Fill missing industry-month observations with zero
monthly_industry_counts = monthly_industry_counts.reindex(
    complete_index,
    fill_value=0
)

# Calculate monthly year-over-year changes
monthly_industry_counts["monthly_change"] = (
    monthly_industry_counts[2026]
    - monthly_industry_counts[2025]
)

# Identify months with positive changes
monthly_industry_counts["increased"] = (
    monthly_industry_counts["monthly_change"] > 0
)

# Count months with increases for each industry
persistence_summary = (
    monthly_industry_counts
    .groupby(level=0)["increased"]
    .sum()
    .reset_index(name="months_increased")
)

# Calculate persistence percentage
persistence_summary["persistence_pct"] = (
    persistence_summary["months_increased"] / 10 * 100
).round(2)

# Rank industries by persistence
persistence_summary = persistence_summary.sort_values(
    "months_increased",
    ascending=False
)

# Validate monthly coverage
monthly_coverage = (
    monthly_industry_counts
    .groupby(level=0)
    .size()
)

assert (monthly_coverage == 10).all(), (
    "Some industries do not have complete monthly coverage."
)

print("Industries checked:", len(monthly_coverage))
print("Industries with incomplete coverage:",
      (monthly_coverage != 10).sum())

# Display final persistence results
display(persistence_summary)
```

### 4.9 Recent Industry Monitoring Priority Matrix

The monitoring framework combines rule-based classification with supporting industry diagnostics to identify areas requiring further investigation.

**Stage 1 - Rule-Based Classification**

Industries are assigned to monitoring categories using two indicators: absolute year-over-year appointment change and monthly persistence.

The rules are applied in the following order:

1. **High-volume increase:** At least 100 additional appointments.
2. **Persistent increase:** A positive net increase, with increases recorded in at least 6 of 10 corresponding months.
3. **Emerging increase:** A positive net increase that does not meet the previous thresholds.
4. **Stable or declining:** No positive net increase.

The comparison covers January-October 2025 and January-October 2026.

The thresholds of 100 appointments and 6 months are project-defined screening criteria, not statistically calibrated credit risk thresholds.

**Stage 2 - Supporting Diagnostic Assessment**

The classification is interpreted alongside additional evidence:

- **Historical deviation:** Whether recent appointment activity differs from the industry's historical baseline.
- **Industry share shift:** Whether the industry accounts for a larger or smaller proportion of national appointments.
- **Event concentration:** Whether unusually large changes are influenced by concentrated appointment events, where investigated.

These indicators provide context but do not directly determine the assigned monitoring category.

**Interpretation**

The monitoring matrix is a screening tool, not a validated credit risk rating system.

A higher monitoring category does not automatically imply a higher probability of default or greater expected credit losses.

Final monitoring recommendations consider both the rule-based category and the supporting diagnostic evidence, with lender-specific exposure and credit performance required before making credit decisions.

```python
# Combine year-over-year and share shift results
monitoring_df = share_shift_df.reset_index().copy()

# Add historical deviation
historical_metrics = historical_counts[
    ["historical_deviation_pct"]
].reset_index()

monitoring_df = monitoring_df.merge(
    historical_metrics,
    on="Industry type (division)",
    how="left",
    validate="one_to_one"
)

# Add persistence indicators
monitoring_df = monitoring_df.merge(
    persistence_summary,
    on="Industry type (division)",
    how="left",
    validate="one_to_one"
)

# Define transparent monitoring categories
def classify_monitoring(row):
    if row["absolute_change"] >= 100:
        return "High-volume increase"

    if row["absolute_change"] > 0 and row["months_increased"] >= 6:
        return "Persistent increase"

    if row["absolute_change"] > 0:
        return "Emerging increase"

    return "Stable or declining"

# Apply classification
monitoring_df["monitoring_category"] = monitoring_df.apply(
    classify_monitoring,
    axis=1
)

# Exclude unknown industry from monitoring rankings
monitoring_df = monitoring_df[
    monitoring_df["Industry type (division)"] != "Unknown"
].copy()

# Display selected indicators
display(
    monitoring_df[
        [
            "Industry type (division)",
            "count_2026",
            "absolute_change",
            "yoy_growth_pct",
            "share_shift_pp",
            "historical_deviation_pct",
            "months_increased",
            "monitoring_category"
        ]
    ].sort_values(
        ["absolute_change", "months_increased"],
        ascending=False
    )
)
```

**Key Findings - Industry Monitoring Priority Matrix**

1. **Construction showed the largest absolute increase:** Construction recorded 3,380 appointments during January-October 2026, an increase of 386 cases (+12.89%) compared with the same period in 2025. Its national appointment share increased by 3.60 percentage points. However, only four of ten months recorded positive year-over-year changes.

2. **Mining showed the most persistent increase:** Mining recorded positive monthly changes in eight of ten months, with overall appointment growth of 25.37%. Although its absolute increase was smaller (+34 cases), the consistency of monthly increases warrants further monitoring.

3. **Arts and Recreation Services also showed persistent increases:** Appointments increased by 22 cases (+11.06%), with positive changes in six of ten months.

4. **Other industries showed mixed signals:** Several industries recorded positive historical deviations despite stable or declining appointment counts in 2026. This demonstrates why historical deviation should not be interpreted independently.

**Business Implication:** Construction is a candidate for volume-focused monitoring, while Mining and Arts and Recreation Services warrant attention for persistent increases. These represent different monitoring needs rather than validated rankings of company-level credit risk.

### 4.10 Construction Monthly Growth Drivers

### 4.12 Construction Sensitivity Analysis

This analysis evaluates how sensitive Construction's year-over-year appointment growth is to the concentrated cluster identified on 25 August 2026.

The official ASIC Series 1 counts are retained unchanged. A separate sensitivity scenario excludes the 540 appointments associated with the same date, appointment type, and appointee.

**Scenarios:**

- **Reported:** All Construction Series 1 appointments during January-October 2025 and 2026.
- **Excluding identified cluster:** The same comparison, excluding the 540 identified appointments from the 2026 period.

The sensitivity scenario is not an alternative official insolvency statistic. It illustrates the extent to which a concentrated group of appointments influences the reported industry trend.

**Limitation:** The shared appointee and effective date do not establish common ownership, a single underlying event, or a common cause of insolvency.

```python
# Get reported Construction counts
construction_2025 = int(
    industry_yearly.loc["Construction", "count_2025"]
)

construction_2026 = int(
    industry_yearly.loc["Construction", "count_2026"]
)

# Standardize appointee names to avoid whitespace issues
cluster_df["appointee_clean"] = (
    cluster_df["Appointee (person or company)"]
    .astype(str)
    .str.strip()
    .str.upper()
)

# Inspect the standardized appointee names
print("Appointee names in the cluster:")
print(cluster_df["appointee_clean"].value_counts())

# Identify appointments linked to the dominant appointee
cluster_mask = (
    (cluster_df["Appointment type"] == "Voluntary administration") &
    (cluster_df["appointee_clean"] == "SCOTT, ANDREW JOHN")
)

cluster_count = int(cluster_mask.sum())

# Validate the cluster before calculating the scenario
print("\nIdentified cluster appointments:", cluster_count)

assert cluster_count == 540, (
    "Cluster count does not match Section 4.11. "
    "Check the appointment type and appointee values."
)

# Calculate reported scenario
reported_change = construction_2026 - construction_2025
reported_growth = (
    reported_change / construction_2025 * 100
)

# Calculate sensitivity scenario
adjusted_2026 = construction_2026 - cluster_count
adjusted_change = adjusted_2026 - construction_2025
adjusted_growth = (
    adjusted_change / construction_2025 * 100
)

# Create comparison table
sensitivity_summary = pd.DataFrame({
    "scenario": [
        "Reported ASIC Series 1",
        "Excluding identified cluster"
    ],
    "count_2025": [
        construction_2025,
        construction_2025
    ],
    "count_2026": [
        construction_2026,
        adjusted_2026
    ],
    "absolute_change": [
        reported_change,
        adjusted_change
    ],
    "yoy_growth_pct": [
        round(reported_growth, 2),
        round(adjusted_growth, 2)
    ]
})

display(sensitivity_summary)
```

Excluding the 540 identified appointments reduces the January-October 2026 Construction count from 3,380 to 2,840, while the 2025 comparison count remains unchanged at 2,994.

### 4.13 Integrated Industry Monitoring Framework

The integrated monitoring framework combines recent insolvency indicators, historical industry trends and event-level investigation results to support industry-level credit risk monitoring.

The objective is to identify appropriate monitoring actions and assess whether incorporating historical evidence changes recommendations derived from recent appointment activity alone.

**Methodology**

The framework consists of three analytical components:

**1. Recent Monitoring Classification**

The monitoring categories established in Section 4.9 provide the baseline assessment of recent industry appointment activity.

**2. Historical Monitoring Signals**

Four historical signals are derived from the multi-year, rolling trend and historical share analyses.

| Historical Signal | Definition |
|---|---|
| Above-national growth | Industry growth exceeds national growth between January-October 2022 and January-October 2026 |
| Consistent annual growth | Appointment counts increase in all four consecutive year-over-year comparisons |
| Rising rolling trend | Rolling 12-month appointment counts increase relative to both six and twelve months earlier |
| Expanding industry share | The industry's share of national rolling appointments increases between June 2022 and October 2026 |

**3. Integrated Monitoring Decision Rules**

Recommended actions are assigned using the following rules in descending order of precedence. The first satisfied rule determines the recommended action.

| Rule | Condition | Recommended Action |
|---|---|---|
| 1 | High-volume increase | Targeted investigation |
| 2 | Persistent increase | Enhanced monitoring |
| 3 | Above-national growth, rising rolling trend and expanding industry share | Enhanced monitoring |
| 4 | Above-national growth, consistent annual growth and expanding industry share | Historical trend review |
| 5 | Emerging increase | Emerging trend watch |
| 6 | None of the preceding conditions are satisfied | Routine monitoring |

The Construction investigation flag and sensitivity result are incorporated into the final monitoring matrix as contextual evidence without modifying the official appointment counts or classification rules.

Historical indicators may capture related aspects of industry appointment activity and should not be interpreted as independent risk factors.

The framework recommends monitoring actions rather than assigning numerical risk scores or estimating probabilities of default. Industries within the same monitoring category may require different levels of attention depending on appointment volumes, portfolio exposure and additional credit risk information.

```python
# Select multi-year growth indicators
historical_monitoring_metrics = multi_year_metrics[
    [
        "long_term_change",
        "long_term_growth_pct",
        "growth_vs_national_pp",
        "positive_growth_years",
        "annual_growth_consistency_pct"
    ]
].copy()

# Integrate rolling trend indicators
historical_monitoring_metrics = historical_monitoring_metrics.join(
    rolling_trend_metrics[
        [
            "latest_12m_count",
            "peak_month",
            "change_from_peak_pct",
            "rolling_6m_change_pct",
            "rolling_12m_change_pct"
        ]
    ],
    how="left",
    validate="one_to_one"
)

# Integrate historical industry share indicators
historical_monitoring_metrics = historical_monitoring_metrics.join(
    historical_share_metrics[
        [
            "latest_share_pct",
            "historical_share_shift_pp",
            "share_6m_shift_pp",
            "share_12m_shift_pp"
        ]
    ],
    how="left",
    validate="one_to_one"
)

# Validate industry coverage and data completeness
assert historical_monitoring_metrics.shape == (19, 14)
assert historical_monitoring_metrics.index.is_unique
assert historical_monitoring_metrics.notna().all().all()

# Sort by latest rolling appointment volume
historical_monitoring_metrics = (
    historical_monitoring_metrics.sort_values(
        "latest_12m_count",
        ascending=False
    )
)

# Display validation results
print("Classified industries:", len(historical_monitoring_metrics))
print("Historical indicators:", len(historical_monitoring_metrics.columns))
print("Missing values:", historical_monitoring_metrics.isna().sum().sum())

# Display integrated historical indicators
display(
    historical_monitoring_metrics
    .round(2)
    .reset_index()
)
```

```python
# Create historical monitoring signals
historical_signals = historical_monitoring_metrics.copy()

# Identify industries growing faster than the national benchmark
historical_signals["above_national_growth"] = (
    historical_signals["growth_vs_national_pp"] > 0
)

# Identify industries with consistent annual growth
historical_signals["consistent_annual_growth"] = (
    historical_signals["positive_growth_years"] == 4
)

# Identify industries with rising rolling appointment totals
historical_signals["rising_rolling_trend"] = (
    (historical_signals["rolling_6m_change_pct"] > 0)
    & (historical_signals["rolling_12m_change_pct"] > 0)
)

# Identify industries with expanding historical shares
historical_signals["expanding_industry_share"] = (
    historical_signals["historical_share_shift_pp"] > 0
)

# Select historical signal columns
historical_signal_columns = [
    "above_national_growth",
    "consistent_annual_growth",
    "rising_rolling_trend",
    "expanding_industry_share"
]

# Create a compact historical signal matrix
historical_signal_matrix = historical_signals[
    historical_signal_columns
].copy()

# Validate industry coverage
assert historical_signal_matrix.shape == (19, 4)
assert historical_signal_matrix.notna().all().all()

# Display historical signal matrix
display(historical_signal_matrix.reset_index())

# Summarise the number of industries meeting each condition
display(
    historical_signal_matrix
    .sum()
    .rename("industry_count")
    .reset_index()
    .rename(columns={"index": "historical_signal"})
)
```

```python
# Select recent monitoring indicators
integrated_monitoring_df = monitoring_df[
    [
        "Industry type (division)",
        "count_2026",
        "absolute_change",
        "yoy_growth_pct",
        "share_shift_pp",
        "historical_deviation_pct",
        "months_increased",
        "monitoring_category"
    ]
].copy()

# Integrate historical monitoring signals
integrated_monitoring_df = integrated_monitoring_df.merge(
    historical_signal_matrix.reset_index(),
    on="Industry type (division)",
    how="left",
    validate="one_to_one"
)

# Validate industry coverage and data completeness
assert integrated_monitoring_df.shape == (19, 12)

assert integrated_monitoring_df[
    "Industry type (division)"
].is_unique

assert integrated_monitoring_df.notna().all().all()

# Sort by recent absolute appointment changes
integrated_monitoring_df = integrated_monitoring_df.sort_values(
    "absolute_change",
    ascending=False
).reset_index(drop=True)

# Display validation results
print("Classified industries:", len(integrated_monitoring_df))
print("Total columns:", len(integrated_monitoring_df.columns))
print("Missing values:", integrated_monitoring_df.isna().sum().sum())

# Display integrated monitoring signals
display(integrated_monitoring_df.round(2))
```

```python
# Create a working dataset for final monitoring actions
final_monitoring_df = integrated_monitoring_df.copy()

# Define monitoring actions using recent and historical evidence
def recommend_monitoring_action(row):

    # Investigate substantial increases in appointment volume
    if row["monitoring_category"] == "High-volume increase":
        return "Targeted investigation"

    # Monitor persistent recent increases
    if row["monitoring_category"] == "Persistent increase":
        return "Enhanced monitoring"

    # Identify industries with aligned historical and rolling signals
    if (
        row["above_national_growth"]
        and row["rising_rolling_trend"]
        and row["expanding_industry_share"]
    ):
        return "Enhanced monitoring"

    # Review sustained historical changes in industry concentration
    if (
        row["above_national_growth"]
        and row["consistent_annual_growth"]
        and row["expanding_industry_share"]
    ):
        return "Historical trend review"

    # Monitor emerging recent increases
    if row["monitoring_category"] == "Emerging increase":
        return "Emerging trend watch"

    # Maintain routine monitoring for remaining industries
    return "Routine monitoring"


# Apply monitoring action rules
final_monitoring_df["recommended_action"] = (
    final_monitoring_df.apply(
        recommend_monitoring_action,
        axis=1
    )
)

# Validate monitoring action coverage
assert len(final_monitoring_df) == 19
assert final_monitoring_df["recommended_action"].notna().all()

# Display recommended monitoring actions
display(
    final_monitoring_df[
        [
            "Industry type (division)",
            "count_2026",
            "absolute_change",
            "yoy_growth_pct",
            "monitoring_category",
            "recommended_action"
        ]
    ].round(2)
)

# Summarise recommended monitoring actions
display(
    final_monitoring_df["recommended_action"]
    .value_counts()
    .rename_axis("recommended_action")
    .reset_index(name="industry_count")
)
```

```python
# Define monitoring actions using recent signals only
recent_action_mapping = {
    "High-volume increase": "Targeted investigation",
    "Persistent increase": "Enhanced monitoring",
    "Emerging increase": "Emerging trend watch",
    "Stable or declining": "Routine monitoring"
}

# Create a comparison dataset
monitoring_action_comparison = final_monitoring_df.copy()

# Establish the recent-only monitoring action
monitoring_action_comparison["recent_only_action"] = (
    monitoring_action_comparison["monitoring_category"]
    .map(recent_action_mapping)
)

# Validate recent-only action coverage
assert monitoring_action_comparison["recent_only_action"].notna().all()

# Identify actions changed by historical evidence
monitoring_action_comparison["action_changed"] = (
    monitoring_action_comparison["recent_only_action"]
    != monitoring_action_comparison["recommended_action"]
)

# Select industries with revised monitoring actions
revised_monitoring_actions = monitoring_action_comparison[
    monitoring_action_comparison["action_changed"]
].copy()

# Display the contribution of historical evidence
print("Total classified industries:", len(monitoring_action_comparison))
print("Revised monitoring actions:", len(revised_monitoring_actions))

print(
    "Share of industries with revised actions:",
    f"{len(revised_monitoring_actions) / len(monitoring_action_comparison) * 100:.2f}%"
)

display(
    revised_monitoring_actions[
        [
            "Industry type (division)",
            "recent_only_action",
            "recommended_action",
            "above_national_growth",
            "consistent_annual_growth",
            "rising_rolling_trend",
            "expanding_industry_share"
        ]
    ]
)
```

```python
# Add investigation status to the integrated monitoring matrix
final_monitoring_df["investigation_flag"] = (
    "Cluster concentration not investigated"
)

# Identify Construction
construction_mask = (
    final_monitoring_df["Industry type (division)"] == "Construction"
)

# Flag the investigated Construction cluster
final_monitoring_df.loc[
    construction_mask,
    "investigation_flag"
] = "Concentrated cluster - sensitivity confirmed"

# Create a separate column for sensitivity results
final_monitoring_df["sensitivity_yoy_pct"] = float("nan")

# Retrieve the Construction sensitivity result
construction_sensitivity = sensitivity_summary.loc[
    sensitivity_summary["scenario"] == "Excluding identified cluster",
    "yoy_growth_pct"
].iloc[0]

# Assign the sensitivity result to Construction only
final_monitoring_df.loc[
    construction_mask,
    "sensitivity_yoy_pct"
] = construction_sensitivity

# Add the recent-only monitoring action for comparison
final_monitoring_df["recent_only_action"] = (
    final_monitoring_df["monitoring_category"]
    .map(recent_action_mapping)
)

# Identify actions changed by historical evidence
final_monitoring_df["action_changed"] = (
    final_monitoring_df["recent_only_action"]
    != final_monitoring_df["recommended_action"]
)

# Validate the consolidated monitoring matrix
assert len(final_monitoring_df) == 19
assert final_monitoring_df["Industry type (division)"].is_unique
assert final_monitoring_df["recommended_action"].notna().all()
assert final_monitoring_df["recent_only_action"].notna().all()
assert final_monitoring_df["sensitivity_yoy_pct"].notna().sum() == 1

# Display validation results
print("Classified industries:", len(final_monitoring_df))
print("Total columns:", len(final_monitoring_df.columns))
print("Revised monitoring actions:", final_monitoring_df["action_changed"].sum())

# Display the consolidated final monitoring matrix
display(
    final_monitoring_df[
        [
            "Industry type (division)",
            "count_2026",
            "absolute_change",
            "yoy_growth_pct",
            "monitoring_category",
            "recommended_action",
            "action_changed",
            "investigation_flag",
            "sensitivity_yoy_pct"
        ]
    ].round(2)
)
```

## 6. Limitations & Conclusion

### 6.1 Analytical Limitations

**1. Appointment counts are not insolvency rates**

ASIC Series 1 measures companies entering external administration or having a controller appointed for the first time under the series definition. The dataset does not provide the total population of active companies in each industry. Therefore, appointment counts and industry shares cannot be interpreted as insolvency rates or probabilities of default.

Repeated ACNs may occur within the dataset. Appointment counts should not be interpreted as the number of unique companies across the entire analysis period.

**2. Appointment activity does not directly measure credit losses**

The dataset does not contain outstanding loan balances, lender exposures, recovery amounts or realised credit losses. Higher appointment volumes cannot automatically be interpreted as greater financial losses for creditors.

**3. Historical indicators are sensitive to reference periods**

Multi-year growth and historical share changes depend on the selected reference periods. Industries with relatively small initial appointment counts may exhibit substantial percentage growth despite modest absolute increases. Historical comparisons should therefore be interpreted alongside appointment volumes and national benchmarks.

**4. Rolling trends may obscure short-term movements**

Rolling 12-month totals reduce the influence of monthly fluctuations but may delay the identification of emerging turning points. Consecutive rolling periods also contain overlapping observations and should not be treated as independent evidence.

**5. Concentrated appointment events can influence industry trends**

The Construction investigation demonstrates that concentrated appointment activity can materially influence aggregate industry indicators. The sensitivity scenario is illustrative and does not establish that the identified records are invalid. Industries not investigated for event concentration should not be assumed to be free of similar events.

**6. Monitoring rules are not externally validated**

The monitoring categories, historical signals and recommended actions are rule-based analytical indicators developed for this project. Their thresholds have not been calibrated against observed defaults or credit losses.

Historical indicators may capture related aspects of the same underlying appointment trends and should not be treated as independent risk factors.

Changes in recommended actions demonstrate the influence of historical evidence, rather than improved predictive accuracy.

### 6.2 Conclusion

This project analysed 58,115 Australian corporate insolvency appointment records across 64 reporting months from July 2021 to October 2026, using ASIC Series 1 data to develop an industry-level monitoring framework relevant to credit risk analysis.

National insolvency activity remains historically elevated despite a modest year-over-year decline in 2026. Multi-year growth, rolling appointment trends and historical share evolution reveal increasingly differentiated industry trajectories that cannot be fully understood through recent year-over-year comparisons alone.

Health Care and Social Assistance exhibits sustained historical growth and expanding national appointment share, while Mining has reached its highest observed rolling appointment count. In contrast, Accommodation and Food Services demonstrates a sustained decline from its historical peak.

The Construction investigation highlights the importance of examining exceptional appointment concentrations before interpreting aggregate increases as broad-based industry deterioration. The sensitivity scenario demonstrates how concentrated appointment activity can materially influence industry monitoring signals.

By integrating recent indicators with historical evidence, the framework identifies one industry for targeted investigation, four for enhanced monitoring, one for historical trend review, one for emerging trend watch and 12 for routine monitoring.

Historical evidence changes the recommended monitoring actions for three of the 19 classified industries (15.79%), demonstrating its practical contribution to the monitoring framework.

**Final Takeaway:** Effective industry distress monitoring requires more than identifying which industries have the highest insolvency appointment counts or the fastest recent growth. Combining long-term trends, recent movements, relative concentration and event-level diagnostics provides a more comprehensive basis for allocating monitoring resources. These insights should complement, rather than replace, borrower-level financial assessments and portfolio-specific credit risk information.

### 7.2 Full Industry Monitoring Validation

This section independently recalculates year-over-year growth, industry share shifts, and monthly persistence for all industries included in the final monitoring matrix.

Validation covers all 19 classified industries. All calculations use January-October 2025 and January-October 2026.

```python
# Prepare independent validation dataset
validation_df = clean_df[
    (clean_df["calendar_year"].isin([2025, 2026])) &
    (clean_df["calendar_month"] <= 10)
].copy()

# Independently calculate national appointment totals
national_totals = (
    validation_df
    .groupby("calendar_year")
    .size()
)

# Verify national totals against reconciled results
assert national_totals.loc[2025] == 12272
assert national_totals.loc[2026] == 12073

industry_column = "Industry type (division)"

# Independently calculate industry counts for both years
validation_counts = (
    validation_df
    .groupby([industry_column, "calendar_year"])
    .size()
    .unstack(fill_value=0)
    .reindex(columns=[2025, 2026], fill_value=0)
)

# Calculate YoY growth and industry share shifts
validation_counts["calculated_yoy_pct"] = (
    (validation_counts[2026] - validation_counts[2025])
    / validation_counts[2025] * 100
)

validation_counts["calculated_share_shift_pp"] = (
    validation_counts[2026] / national_totals.loc[2026] * 100
    - validation_counts[2025] / national_totals.loc[2025] * 100
)

# Independently calculate monthly persistence
monthly_validation = (
    validation_df
    .groupby([industry_column, "calendar_year", "calendar_month"])
    .size()
    .unstack("calendar_year", fill_value=0)
    .reindex(columns=[2025, 2026], fill_value=0)
)

# Include zero-count months for each industry
all_months = pd.MultiIndex.from_product(
    [validation_counts.index, range(1, 11)],
    names=[industry_column, "calendar_month"]
)

monthly_validation = monthly_validation.reindex(
    all_months,
    fill_value=0
)

calculated_persistence = (
    (monthly_validation[2026] > monthly_validation[2025])
    .groupby(level=0)
    .sum()
)

validation_counts["calculated_months_increased"] = calculated_persistence

# Compare independently calculated values with the final matrix
audit_df = final_monitoring_df.set_index(industry_column).join(
    validation_counts[
        [
            "calculated_yoy_pct",
            "calculated_share_shift_pp",
            "calculated_months_increased"
        ]
    ],
    how="left"
)

audit_df["yoy_match"] = (
    audit_df["yoy_growth_pct"].round(2)
    == audit_df["calculated_yoy_pct"].round(2)
)

audit_df["share_match"] = (
    audit_df["share_shift_pp"].round(2)
    == audit_df["calculated_share_shift_pp"].round(2)
)

audit_df["persistence_match"] = (
    audit_df["months_increased"]
    == audit_df["calculated_months_increased"]
)

audit_df["all_checks_pass"] = (
    audit_df["yoy_match"]
    & audit_df["share_match"]
    & audit_df["persistence_match"]
)

# Display validation results
display(
    audit_df[
        [
            "yoy_match",
            "share_match",
            "persistence_match",
            "all_checks_pass"
        ]
    ].reset_index()
)

print("Industries checked:", len(audit_df))
print("Industries passing all checks:", audit_df["all_checks_pass"].sum())
print("Industries failing:", (~audit_df["all_checks_pass"]).sum())

assert len(audit_df) == 19
assert audit_df["all_checks_pass"].all(), (
    "Some industry monitoring metrics failed validation."
)

print("PASS: All 19 industries validated successfully.")
```
